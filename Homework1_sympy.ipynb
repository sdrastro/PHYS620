## $f - y^{\prime} \frac{\partial f}{\partial y^{\prime}} = C$

```python
r = symbols('r')
phi = Function('phi')(r)
phip = phi.diff(r)
```

```python
f = r**(-2) * sqrt(1 + r**2 * phip**2)
```

```python
EL = calculus.euler_equations(f, phi, r)[0]
```

```python
EL
```

```python
simplify(EL)
```

```python
dsolve(simplify(EL), phi)
```

```python
C = symbols('C', real=True, constant=True)
expr = Eq(r**-2 * (1 + r**2 * phip**2)**(1/2), C)
```

```python
Derivative(expr, phip)
```
